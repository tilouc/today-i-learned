```python
# data structure

# list slicing
amazon_cart = [
    'notebooks',
    'sunglasses',
    'toys',
    'grapes'
]
print(amazon_cart[0:2])
print(amazon_cart[0::2])
print('')

amazon_cart[0] = 'laptop'
new_cart = amazon_cart[1:3]
print(amazon_cart)
print(new_cart)
print('')

new_cart = amazon_cart
amazon_cart[0] = 'desktop'
print(amazon_cart)
print(new_cart)
print('')

new_cart = amazon_cart[:]
new_cart[0] = 'gum'
new_cart2 = amazon_cart[0:3]
print(new_cart)
print(new_cart2)
print(amazon_cart)
```

```python
# list methods
basket = [1,2,3,4,5]
print(len(basket))
print('')

# adding
basket.append(100)
basket.insert(5, 99)
new_list = basket.extend([101])
basket.extend([102])
print(basket)
print(new_list) # returns None
print('')

# removing
basket.pop()
basket.pop(0) # index

basket.remove(101) # value

new_list = basket.pop(4)
new_list = basket.clear()

print(basket)
print(new_list)
```

```python
basket = ['a','x','b','c','d','e','d']

# print(basket.index('d', 0, 4))
print(basket.index('d', 0, len(basket)))
print('')

print('d' in basket)
print('x' in basket)
print('y' in basket)
print('')

print('i' in 'hi my name is ian')
print('')

print(basket.count('d'))
print('')

basket.sort()

new_basket = basket.copy()
new_basket = basket[:]
new_basket.sort()
basket.reverse()

print(basket)
print(len(basket))
print(new_basket)
print(sorted(basket))
print(basket[::-1])
```
