```python
# dictionary
dictionary = {'a': 1,'b': 2,'x': 3}

dictionary = {
    'a': [1,2,3],
    'b': 'hello',
    'x': True
}

my_list = [
    {
    'a': [1,2,3],
    'b': 'hello',
    'x': True
    },
    {
    'a': [4,5,6],
    'b': 'hello',
    'x': True
    }
]

print(dictionary)
print(dictionary['a'])
print(dictionary['a'][1])
print(dictionary['b'])
print(dictionary['x'])
print(my_list)
print(my_list[0]['a'][2])
```

```python
dictionary = {
    'weapons': [1,2,3],
    'greeting': 'hello',
    'is_Magic': True
}

dictionary = {
    123: [1,2,3],
    '123': [1,2,3],
    '123': 'hello', # overwrite
    True: 'hello',
    # [100]: True, # impossible
    '[100]': True
}

# [0] = # impossible

print(dictionary[123])
print(dictionary['123'])
print(dictionary[True])
print(dictionary['[100]'])
```
