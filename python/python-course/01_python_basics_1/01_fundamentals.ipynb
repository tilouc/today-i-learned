```python
# int and float
print(2 + 4)
print(2 - 4)
print(2 * 4)
print(2 / 4)
print('')

print(type(6))
print(type(2 - 4))
print(type(2 * 4))
print(type(2 / 4)) # 0.5
print('')

print(type(0.00001))
print(type(5.00001))
print(type(0))
print(type(20+1.1))
print(type(9.9+1.1))
print('')

print(9.9+1.1) # 11.0
print('')

print(2 ** 2)
print(2 ** 3) # power
print('')

print(2 // 4)
print(3 // 4)
print(5 // 4) # gets rounded down to an integer
print('')

print(5 % 4)
print(6 % 4)
print(8 % 4) # modulo
```
